### Create a string, sent = ‘Coronavirus Caused Lockdowns Around The World.”

```python
sent = 'Coronavirus Caused Lockdowns Around The World'
```

Use split function to convert the string into a list of words and save it in variable words and print the same

```python


Word = sent.split()
print(Word)
```

Convert each word in the list to lower case and store it in variable words_lower.
Print words_lower

```python

words_lower = list(map(str.lower, Word))
print(words_lower)
```

Check whether ‘country’ is in the list

```python
if 'country' in words_lower:
    print("Yes, 'country' is found!")
else:
    print("No, 'country' is not found!")
```

Remove the element ‘the’ from the list and print the list.

```python
words_lower.remove('the')
print(words_lower)
```

Select the first 4 words from the list words_lower using slicing and store them in a new variable x4

```python
x4 = words_lower[:4]
```

```python
# print x4

print(x4)
```

Convert the list of elements to single string using join function and print it

```python

sentence = ' '.join(x4)
print(sentence)
```

### Create colors = [‘red’,’blue’,’orange’], and fruits = [‘orange’,’grapes’,’apples’]

```python
colors = ['red', 'blue','orange']
fruits = ['orange','grape','apple']
```

Print color and fruits

```python

print("Colors are:", colors, "and fruits are:", fruits)
```

Create colors_set, and fruits_set. (using set() ) and print them

```python
colors_set = set(colors)
fruits_set = set(fruits)
print(colors_set,"&", fruits_set)
```

Find the union of both the sets.

```python

print("Union of colors_set and fruits_set:", colors_set.union(fruits_set))

```

Find the intersection of both the sets

```python


print("Intersection of colors_set and fruits_set:", colors_set.intersection(fruits_set))
```

Find the elements which are Fruits but not colors (using set.difference() )

```python


print("Difference of fruits_set and colors_set:", fruits_set.difference(colors_set))
```

```python
print("Difference of colors_set and fruits_set:", colors_set.difference(fruits_set))
```
